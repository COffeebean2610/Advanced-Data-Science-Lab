```python
# ============================================================
# PRACTICAL NO. 04
# OBJECT DETECTION AND IMAGE ANALYTICS USING YOLO AND OPENCV
# ============================================================

import cv2
import urllib.request
from ultralytics import YOLO

# ============================================================
# 1. DOWNLOAD SAMPLE IMAGE AUTOMATICALLY
# ============================================================

image_url = "https://ultralytics.com/images/bus.jpg"
image_path = "sample_image.jpg"

urllib.request.urlretrieve(image_url, image_path)

print("Sample image downloaded successfully!")

# ============================================================
# 2. LOAD PRE-TRAINED YOLO MODEL
# ============================================================

model = YOLO("yolov8n.pt")

print("YOLO model loaded successfully!")

# ============================================================
# 3. READ IMAGE USING OPENCV
# ============================================================

image = cv2.imread(image_path)

if image is None:
    print("Error: Image could not be loaded!")
    exit()

# ============================================================
# 4. RUN YOLO OBJECT DETECTION
# ============================================================

results = model(image, verbose=False)

# ============================================================
# 5. OBJECT COUNTERS
# ============================================================

person_count = 0
vehicle_count = 0

vehicle_classes = [
    "car",
    "bus",
    "truck",
    "motorcycle"
]

# ============================================================
# 6. PROCESS DETECTIONS
# ============================================================

for result in results:

    boxes = result.boxes

    for box in boxes:

        # Get class ID
        cls_id = int(box.cls[0])

        # Get class label
        label = model.names[cls_id]

        # Get confidence score
        confidence = float(box.conf[0])

        # Get bounding box coordinates
        x1, y1, x2, y2 = map(
            int,
            box.xyxy[0]
        )

        # ====================================================
        # COUNT PERSONS
        # ====================================================

        if label == "person":
            person_count += 1

        # ====================================================
        # COUNT VEHICLES
        # ====================================================

        if label in vehicle_classes:
            vehicle_count += 1

        # ====================================================
        # DRAW BOUNDING BOX
        # ====================================================

        cv2.rectangle(
            image,
            (x1, y1),
            (x2, y2),
            (0, 255, 0),
            2
        )

        # ====================================================
        # DISPLAY LABEL AND CONFIDENCE
        # ====================================================

        text = f"{label} {confidence:.2f}"

        cv2.putText(
            image,
            text,
            (x1, y1 - 10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0, 255, 0),
            2
        )

# ============================================================
# 7. DISPLAY DETECTION COUNTS
# ============================================================

cv2.putText(
    image,
    f"Persons: {person_count}",
    (20, 30),
    cv2.FONT_HERSHEY_SIMPLEX,
    0.8,
    (0, 0, 255),
    2
)

cv2.putText(
    image,
    f"Vehicles: {vehicle_count}",
    (20, 60),
    cv2.FONT_HERSHEY_SIMPLEX,
    0.8,
    (0, 0, 255),
    2
)

# ============================================================
# 8. SAVE OUTPUT IMAGE
# ============================================================

output_path = "yolo_detection_output.jpg"

cv2.imwrite(
    output_path,
    image
)

print("\nDetection completed successfully!")

print("Persons Detected:", person_count)
print("Vehicles Detected:", vehicle_count)

print("Output saved as:", output_path)

# ============================================================
# 9. DISPLAY OUTPUT
# ============================================================

cv2.imshow(
    "YOLO Object Detection",
    image
)

cv2.waitKey(0)

cv2.destroyAllWindows()

# ============================================================
# END OF PROGRAM
# ============================================================
```
